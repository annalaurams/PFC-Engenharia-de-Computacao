```python
import pandas as pd

# 1. Carregamento dos dados
path = r'/home/anna/tcc/Tratamento/Mamografia/tabela_mamografia.csv'
df = pd.read_csv(path)

# --- TRATAMENTO PRÉVIO (Importante para não dar erro 'NaN') ---
# Transformamos tudo que não for o nome do município em número, tratando vazios como 0
cols_numericas = df.columns.drop('Municípios') # Ajuste se a coluna tiver outro nome como 'MUNICÍPIOS'
df[cols_numericas] = df[cols_numericas].apply(pd.to_numeric, errors='coerce').fillna(0)

# --- PRIMEIRA SOMA: Todas as idades do ano de 2025 ---
# Aqui filtramos todas as colunas que começam com '2025_'
colunas_2025 = [col for col in df.columns if col.startswith('2025_')]
soma_total_2025 = df[colunas_2025].sum().sum()

# --- SEGUNDA SOMA: Todos os anos (2013-2025) da faixa 30-39 ---
# Aqui filtramos todas as colunas que terminam com '_30-39'
colunas_30_39 = [f"{ano}_30-39" for ano in range(2013, 2026)]
# Verificamos se as colunas realmente existem no seu CSV antes de somar
colunas_existentes_30_39 = [c for c in colunas_30_39 if c in df.columns]
soma_historica_30_39 = df[colunas_existentes_30_39].sum().sum()

# --- EXIBIÇÃO DOS RESULTADOS ---
print("="*50)
print("RELATÓRIO DE MAMOGRAFIAS - MINAS GERAIS")
print("="*50)
print(f"1. Total de exames em 2025 (todas as idades): {soma_total_2025:,.0f}")
print(f"2. Total de exames (2013-2025) na faixa 30-39: {soma_historica_30_39:,.0f}")
print("="*50)

# Caso queira ver a soma detalhada por ano na faixa 30-39:
print("\nDetalhamento anual da faixa 30-39:")
print(df[colunas_existentes_30_39].sum())
```
