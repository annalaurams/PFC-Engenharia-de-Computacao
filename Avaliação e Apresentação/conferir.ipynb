```python
import pandas as pd
from pathlib import Path

PASTA_TAXAS = Path("/home/anna/tcc/Seleção e Transformação/População/Taxa")

arquivo = PASTA_TAXAS / "tabela_mamografias_por_faixa_etária_taxa.csv"

tabela = pd.read_csv(arquivo)

REGIONAL_ALVO = "Passos"
ANO_ALVO = 2017

FAIXAS_ETARIAS = ("0-9", "10-19", "20-29", "30-39", "40-49", "50-59", "60-69", "70 OU MAIS")

# Filtra Passos
passos = tabela[
    tabela["REGIONAL DE SAÚDE"].str.strip().str.lower() == REGIONAL_ALVO.lower()
].copy()

# Colunas das faixas etárias do ano (sem TOTAL -- TOTAL já é a soma delas,
# incluir as duas juntas dobraria o resultado)
colunas_faixas = [f"{ANO_ALVO}_{faixa}" for faixa in FAIXAS_ETARIAS]

# Mostra os valores de cada faixa etária e o TOTAL do ano
print("Mamografias de Passos em 2017:")
print(passos[["REGIONAL DE SAÚDE"] + colunas_faixas + [f"{ANO_ALVO}_TOTAL"]])

# Soma das faixas etárias (deve bater com a coluna TOTAL)
total_2017 = passos[colunas_faixas].sum(axis=1).sum()

print(f"\nTotal de mamografias em Passos em {ANO_ALVO}: {total_2017}")
```

```python
import re

COLUNA_REGIONAL = "REGIONAL DE SAÚDE"
FAIXAS_ETARIAS = ("0-9", "10-19", "20-29", "30-39", "40-49", "50-59", "60-69", "70 OU MAIS")

colunas_por_faixa = {
    faixa: [c for c in df.columns if re.fullmatch(rf"\d{{4}}_{re.escape(faixa)}", c)]
    for faixa in FAIXAS_ETARIAS
}
colunas_total = [c for c in df.columns if re.fullmatch(r"\d{4}_TOTAL", c)]

# Somatório de TODOS os anos (1970-2025) de cada coluna, por Regional --
# min_count=1 preserva NaN se a Regional não tiver nenhum valor válido
# em nenhum ano daquela faixa (em vez de virar 0 silenciosamente).
somatorio = pd.DataFrame({
    COLUNA_REGIONAL: df[COLUNA_REGIONAL],
    **{faixa: df[colunas].sum(axis=1, min_count=1) for faixa, colunas in colunas_por_faixa.items()},
    "TOTAL (sem considerar faixa etária)": df[colunas_total].sum(axis=1, min_count=1),
})

# Conferência: a soma das 8 faixas etárias deve bater com a coluna TOTAL,
# regional a regional.
soma_das_faixas = somatorio[list(FAIXAS_ETARIAS)].sum(axis=1)
somatorio["SOMA DAS FAIXAS (conferência)"] = soma_das_faixas
somatorio["DIFERENÇA (TOTAL - SOMA DAS FAIXAS)"] = (
    somatorio["TOTAL (sem considerar faixa etária)"] - soma_das_faixas
)

print("Somatório de todos os anos (1970-2025), por Regional -- população por faixa etária e total:")
somatorio
```
