# Conferência — Ituiutaba e Pirapora em 2023 (Mortalidade)

```python
import pandas as pd

df = pd.read_csv("Mortalidade/tabela_mortes_por_faixa_etária.csv")
colunas_2023 = [c for c in df.columns if c.startswith("2023_")]

regionais = ["Ituiutaba", "Pirapora"]
tabela = df[df["REGIONAL DE SAÚDE"].isin(regionais)]
tabela[["MUNICÍPIO", "REGIONAL DE SAÚDE"] + colunas_2023]
```

### Soma por Regional em 2023 (como fica depois da agregação)

```python
tabela.groupby("REGIONAL DE SAÚDE")[colunas_2023].sum(min_count=1)
```
